```python
# Convert columns into appropriate data types
df["WEEK"] = pd.to_datetime(df["WEEK"], errors="coerce")

for col in ["EVENTS", "FATALITIES", "POPULATION_EXPOSURE"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Extract year
df["YEAR"] = df["WEEK"].dt.year

# Identify top 5 countries based on fatalities
top5_countries = (
    df.groupby("COUNTRY")["FATALITIES"]
    .sum()
    .nlargest(5)
    .index.tolist()
)

print("Top 5 Countries:", top5_countries)

# Filter dataset
top5_df = df[df["COUNTRY"].isin(top5_countries)].copy()
```

## 1. Identification of Countries with the Highest Fatalities

To identify the countries most severely affected by conflict-related events, the total fatalities were aggregated for each country. The top five countries were selected based on their cumulative fatalities across the dataset.

A horizontal bar chart is used to compare the countries and highlight those experiencing the highest human impact.

**Objective:**
- Identify the five countries with the highest recorded fatalities.
- Compare the magnitude of fatalities across countries.
- Establish a basis for further country-level conflict analysis.

```python
# Convert WEEK into datetime
df["WEEK"] = pd.to_datetime(df["WEEK"], errors="coerce")

# Extract year
df["YEAR"] = df["WEEK"].dt.year

# Identify top 3 event types by total fatalities
top_event_types = (
    df.groupby("EVENT_TYPE")["FATALITIES"]
    .sum()
    .nlargest(3)
    .index.tolist()
)

print("Top 3 event types:", top_event_types)

# Filter high-impact events
impact_df = df[
    df["EVENT_TYPE"].isin(top_event_types)
].copy()

# Aggregate yearly data for all countries
yearly_impact = (
    impact_df.groupby(
        ["YEAR", "COUNTRY"],
        as_index=False
    )
    .agg(
        EVENTS=("EVENTS", "sum"),
        FATALITIES=("FATALITIES", "sum")
    )
)

print(yearly_impact.head())
```

```python
# Calculate top 5 countries
country_fatalities = (
    df.groupby("COUNTRY")["FATALITIES"]
    .sum()
    .nlargest(5)
    .reset_index()
)

# Create graph
country_fig = px.bar(
    country_fatalities,
    x="FATALITIES",
    y="COUNTRY",
    orientation="h",
    text="FATALITIES",
    color="FATALITIES",
    color_continuous_scale="Reds",
    title="Top 5 Countries by Total Fatalities",
    template="plotly_white"
)

country_fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

country_fig.update_layout(
    title_x=0.5,
    height=500,
    showlegend=False,
    xaxis_title="Total Fatalities",
    yaxis_title="Country"
)

country_fig.show(renderer="vscode")

country_fig.write_html(
    "../reports/top5_countries_fatalities.html"
)
```

## 2. Relationship Between Population Exposure and Fatalities

This analysis explores the relationship between population exposure and conflict-related fatalities across the five most affected countries.

A bubble scatter plot is used, where:
- X-axis represents median population exposure.
- Y-axis represents total fatalities.
- Bubble size represents total recorded events.
- Bubble colour distinguishes countries.

This visualization helps examine whether countries with higher population exposure also experience higher fatalities and highlights differences in conflict impact.

**Objective:**
- Examine the relationship between population exposure and fatalities.
- Compare the human impact across countries.
- Identify countries with disproportionately high fatalities relative to their typical population exposure.

*Note: Median record-level population exposure is used to avoid simply adding potentially repeated population exposure values. It should not be interpreted as the country's total population.*

```python
country_summary = (
    top5_df.groupby("COUNTRY")
    .agg(
        TOTAL_EVENTS=("EVENTS", "sum"),
        TOTAL_FATALITIES=("FATALITIES", "sum"),
        MEDIAN_POPULATION_EXPOSURE=("POPULATION_EXPOSURE", "median")
    )
    .reset_index()
)

country_summary = country_summary.dropna(
    subset=["MEDIAN_POPULATION_EXPOSURE", "TOTAL_FATALITIES"]
)

fig = px.scatter(
    country_summary,
    x="MEDIAN_POPULATION_EXPOSURE",
    y="TOTAL_FATALITIES",
    size="TOTAL_EVENTS",
    color="COUNTRY",
    text="COUNTRY",
    hover_name="COUNTRY",
    size_max=55,
    title="Population Exposure vs Fatalities Across Top 5 Countries",
    labels={
        "MEDIAN_POPULATION_EXPOSURE": "Median Population Exposure",
        "TOTAL_FATALITIES": "Total Fatalities",
        "TOTAL_EVENTS": "Total Events"
    },
    template="plotly_white"
)

fig.update_traces(
    textposition="top center",
    marker=dict(opacity=0.8, line=dict(width=1, color="white"))
)

fig.update_layout(
    title_x=0.5,
    height=600,
    font=dict(family="Arial", size=13),
    xaxis=dict(showgrid=True, gridcolor="#EAEAEA"),
    yaxis=dict(showgrid=True, gridcolor="#EAEAEA")
)

fig.show()
```

```python
fig.write_html(
    "../reports/population_vs_fatalities.html",
    include_plotlyjs=True
)
```

## 3. Analysis of Fatalities by Event Type

This analysis examines the contribution of different event categories to total conflict-related fatalities across the selected five countries.

Fatalities are aggregated according to the main event categories, allowing us to identify which types of conflict events have the greatest human impact.

A horizontal bar chart is used to rank event categories by their total fatalities.

**Objective:**
- Identify the event categories responsible for the highest fatalities.
- Compare the severity of different event categories.
- Understand which types of conflict events contribute most to the overall human impact.

```python
event_fatalities = (
    top5_df.groupby("EVENT_TYPE", as_index=False)
    ["FATALITIES"].sum()
    .sort_values("FATALITIES", ascending=True)
)

fig = px.bar(
    event_fatalities,
    x="FATALITIES",
    y="EVENT_TYPE",
    orientation="h",
    text="FATALITIES",
    color="FATALITIES",
    color_continuous_scale="Reds",
    title="Fatalities by Event Type Across Top 5 Countries",
    template="plotly_white"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    height=480,
    showlegend=False,
    xaxis_title="Total Fatalities",
    yaxis_title="Event Type",
    font=dict(family="Arial", size=13)
)

fig.show()
```

```python
fig.write_html(
    "../reports/event_type_fatalities.html",
    include_plotlyjs=True
)
```

## 4. Country-wise Distribution of Fatalities by Event Type

This analysis compares the contribution of different event categories to fatalities across the five selected countries.

A stacked bar chart is used to represent total fatalities for each country, with individual event categories displayed as different segments.

**Objective:**
- Compare fatality patterns across the selected countries.
- Identify the dominant event categories within each country.
- Examine similarities and differences in conflict-related human impact.

```python
country_event = (
    top5_df.groupby(
        ["COUNTRY", "EVENT_TYPE"],
        as_index=False
    )["FATALITIES"].sum()
)

fig = px.bar(
    country_event,
    x="COUNTRY",
    y="FATALITIES",
    color="EVENT_TYPE",
    barmode="stack",
    title="Country-wise Fatalities by Event Type",
    template="plotly_white",
    color_discrete_sequence=px.colors.qualitative.Bold
)

fig.update_layout(
    title_x=0.5,
    height=550,
    xaxis_title="Country",
    yaxis_title="Total Fatalities",
    legend_title="Event Type",
    hovermode="x unified",
    font=dict(family="Arial", size=13)
)

fig.show()
```

```python
fig.write_html(
    "../reports/event_type_fatalities.html",
    include_plotlyjs=True
)
```

## 5. Heatmap of Fatalities Across Countries and Event Types

A heatmap is used to compare fatalities across countries and event categories simultaneously.

Each row represents a country, while each column represents an event category. The colour intensity indicates the number of fatalities associated with each combination.

**Objective:**
- Identify the most fatal event categories in individual countries.
- Compare event severity across countries.
- Detect differences in the distribution of conflict-related fatalities.

Darker cells represent higher fatality counts.

```python
heatmap_data = country_event.pivot(
    index="COUNTRY",
    columns="EVENT_TYPE",
    values="FATALITIES"
).fillna(0)

fig = px.imshow(
    heatmap_data,
    text_auto=".2s",
    color_continuous_scale="Reds",
    aspect="auto",
    title="Fatality Distribution Across Countries and Event Types",
    labels={
        "x": "Event Type",
        "y": "Country",
        "color": "Fatalities"
    },
    template="plotly_white"
)

fig.update_layout(
    title_x=0.5,
    height=500,
    font=dict(family="Arial", size=13)
)

fig.show()
```

```python
fig.write_html(
    "../reports/country_event_heatmap.html",
    include_plotlyjs=True
)
```

## 6. Detailed Analysis of Fatalities by Event and Sub-Event Types

This analysis explores conflict-related fatalities at a more detailed level by examining both the main event categories and their corresponding sub-event categories.

A hierarchical treemap is used to visualize the distribution of fatalities. The main event categories form the parent groups, while sub-event categories represent the individual components.

The size of each rectangle corresponds to the number of fatalities.

**Objective:**
- Identify the sub-event categories associated with the highest fatalities.
- Understand which specific incident types contribute most to the main event categories.
- Explore the detailed composition of conflict-related human impact.

```python
sub_event_data = (
    top5_df.groupby(
        ["EVENT_TYPE", "SUB_EVENT_TYPE"],
        as_index=False
    )["FATALITIES"].sum()
)

fig = px.treemap(
    sub_event_data,
    path=["EVENT_TYPE", "SUB_EVENT_TYPE"],
    values="FATALITIES",
    color="FATALITIES",
    color_continuous_scale="Reds",
    title="Fatalities by Event Type and Sub-Event Type",
    template="plotly_white"
)

fig.update_layout(
    title_x=0.5,
    height=650,
    font=dict(family="Arial", size=13)
)

fig.show()
```

```python
fig.write_html(
    "../reports/sub_event_treemap.html",
    include_plotlyjs=True
)
```

## 7. Yearly Trends in Fatalities Across the Most Affected Countries

This analysis examines how conflict-related fatalities have changed over time across the five countries with the highest cumulative fatalities.

The weekly date field is converted into yearly observations, and fatalities are aggregated for each country and year.

A multi-line chart is used to compare annual fatality trends across countries.

**Objective:**
- Identify years with significant increases in fatalities.
- Compare annual conflict severity across countries.
- Examine changes in fatalities over the observation period.
- Identify countries with consistently high or fluctuating fatality trends.

```python
yearly_fatalities = (
    top5_df.dropna(subset=["YEAR"])
    .groupby(["YEAR", "COUNTRY"], as_index=False)
    ["FATALITIES"].sum()
)

fig = px.line(
    yearly_fatalities,
    x="YEAR",
    y="FATALITIES",
    color="COUNTRY",
    markers=True,
    title="Yearly Fatality Trends Across Top 5 Countries",
    template="plotly_white"
)

fig.update_layout(
    title_x=0.5,
    height=550,
    xaxis_title="Year",
    yaxis_title="Total Fatalities",
    hovermode="x unified",
    font=dict(family="Arial", size=13)
)

fig.show()
```

```python
fig.write_html(
    "../reports/yearly_fatalities.html",
    full_html=True,
    include_plotlyjs=True
)
```

## Yearly Trends of High-Impact Conflict Events Across Countries

This analysis examines the yearly distribution of high-impact conflict events across all countries in the dataset.

To focus on events with the greatest human impact, the three event categories with the highest cumulative fatalities are identified. The dataset is then filtered to include these categories, and the total number of events and fatalities is aggregated annually for each country.

A multi-line time-series chart is used to compare yearly event trends across countries.

**Objectives:**
- Identify the countries experiencing the highest frequency of high-impact events.
- Examine changes in conflict event frequency over the years.
- Compare yearly conflict patterns across countries.
- Understand the relationship between event frequency and fatalities.
- Identify periods of increased conflict activity and human impact.

This visualization provides a comparative overview of high-impact conflict patterns across the countries included in the dataset.

```python
# Identify event types causing the highest fatalities

top_event_types = (
    df.groupby("EVENT_TYPE")["FATALITIES"]
    .sum()
    .nlargest(3)
    .index.tolist()
)

print("Top 3 High-Impact Event Types:")

print(top_event_types)
```

```python
# Convert WEEK into datetime
df["WEEK"] = pd.to_datetime(
    df["WEEK"],
    errors="coerce"
)

# Extract year
df["YEAR"] = df["WEEK"].dt.year

# Filter top 3 high-impact event types
impact_df = df[
    df["EVENT_TYPE"].isin(top_event_types)
].copy()

# Aggregate yearly events and fatalities
yearly_impact = (
    impact_df.groupby(
        ["YEAR", "COUNTRY"],
        as_index=False
    )
    .agg(
        EVENTS=("EVENTS", "sum"),
        FATALITIES=("FATALITIES", "sum")
    )
)

yearly_impact = yearly_impact.sort_values(
    ["YEAR", "COUNTRY"]
)

display(yearly_impact.head(10))
```

```python
import plotly.express as px

fig = px.line(
    yearly_impact,
    x="YEAR",
    y="EVENTS",
    color="COUNTRY",
    markers=True,
    hover_data={
        "YEAR": True,
        "COUNTRY": True,
        "EVENTS": ":,",
        "FATALITIES": ":,",
    },
    title="Yearly Trends of High-Impact Conflict Events Across All Countries",
    template="plotly_white"
)

fig.update_traces(
    line=dict(width=2),
    marker=dict(size=6)
)

fig.update_layout(
    title=dict(
        x=0.5,
        font=dict(size=20, color="#183153")
    ),

    height=700,

    xaxis_title="Year",
    yaxis_title="Total Number of Events",

    hovermode="x unified",

    legend=dict(
        title="Countries",
        orientation="v",
        yanchor="top",
        y=1,
        xanchor="left",
        x=1.02
    ),

    font=dict(
        family="Arial",
        size=13,
        color="#263445"
    ),

    plot_bgcolor="white",
    paper_bgcolor="white",

    xaxis=dict(
        showgrid=True,
        gridcolor="#E8EDF3",
        dtick=1
    ),

    yaxis=dict(
        showgrid=True,
        gridcolor="#E8EDF3"
    ),

    margin=dict(
        l=80,
        r=180,
        t=100,
        b=80
    )
)

fig.show()
```

```python
fig.write_html(
    "../reports/yearly_high_impact_events_all_countries.html",
    include_plotlyjs=True
)
```
